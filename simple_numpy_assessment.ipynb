```python
# 1️⃣ Create a NumPy array from this list:
# [4, 8, 12, 16, 20]
# → Use np.array()

import numpy as np
np.array([4, 8, 12, 16, 20])
```

```python
# 2️⃣ Generate an array of numbers from 1 to 30 with a step of 3
# → Use np.arange()

np.arange(1,30,3)
```

```python

np.ones((2,4))
```

```python
# 5️⃣ Reshape this array into 3 rows and 2 columns:
# [1, 2, 3, 4, 5, 6]
# → Use np.reshape()

arr=np.array([1, 2, 3, 4, 5, 6])
arr.reshape(3,2)
```

```python
# 6️⃣ Find the sum, mean, max, and min of the array:
# [10, 20, 30, 40, 50]
# → Use np.sum(), np.mean(), np.max(), np.min()

arr=np.array([10, 20, 30, 40, 50])
np.sum(arr)
```

```python
arr=np.array([10, 20, 30, 40, 50])
np.mean(arr)
```

```python
arr=np.array([10, 20, 30, 40, 50])
np.max(arr)
```

```python
arr=np.array([10, 20, 30, 40, 50])
np.min(arr)
```

```python
# 7️⃣ Sort this array:
# [9, 2, 7, 1, 6]
# → Use np.sort()

arr=np.array([9, 2, 7, 1, 6])
np.sort(arr)
```

```python
# 1️⃣ Generate an array of 5 random integers between 10 and 50
# → Use np.random.randint()

np.random.randint(10,50)
```

```python
# 2️⃣ Generate a 2×3 matrix of random values between 0 and 1
# → Use np.random.rand()

np.random.rand(2,3)
```

```python
# 3️⃣ Combine these two arrays into one:
# a = np.array([1, 2, 3])
# b = np.array([4, 5, 6])
# → Use np.concatenate()

a=np.array([1,2,3])
b=np.array([4,5,6])
np.concatenate([a,b])
```

```python
# 4️⃣ From this array:
# arr = np.array([3, 7, 2, 7, 9, 2, 3])
# Find the unique values.
# → Use np.unique()

arr=np.array([3, 7, 2, 7, 9, 2, 3])
np.unique(arr)
```

```python
# 5️⃣ Using this matrix:
# A = np.array([[1, 2],
#               [3, 4]])
# Calculate its transpose.
# → Use A.T or np.transpose()
A = np.array([[1, 2],
              [3, 4]])
np.transpose(A)
```

```python
# 6️⃣ Perform matrix multiplication:
# A = np.array([[1, 2],
#               [3, 4]])
# B = np.array([[5, 6],
#               [7, 8]])
# → Use np.dot(A, B)
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])
np.dot(A,B)
```

```python
# 7️⃣ Find the indices where elements of this array are greater than 50:
# arr = np.array([12, 67, 45, 89, 23])
# → Use np.where()

arr = np.array([12, 67, 45, 89, 23])
np.where(arr>50)
```
